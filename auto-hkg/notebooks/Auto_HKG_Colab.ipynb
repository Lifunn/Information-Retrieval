```python
# ── 5. Jalankan pipeline ───────────────────────────────────────────
# Tambahkan src/ ke path
import sys
sys.path.insert(0, 'src')

from auto_hkg import AutoHKG, HKGConfig

cfg = HKGConfig(
    provider='groq',
    model='llama-3.3-70b-versatile',
    data_path='data/Knowledge_Base_Update.csv',
    batch_size=10,
    sleep_between_batches=0.5,
)

pipeline = AutoHKG(cfg)
pipeline.run()
```
